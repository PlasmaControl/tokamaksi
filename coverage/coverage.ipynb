# DIII-D coverage

How much of DIII-D the Tokamak-SI labeling pass has processed, over every shot number
from 100000 to the latest in the listing. `processed.txt` (next to this notebook) lists
the processed shots as ranges. A shot skipped on purpose (no plasma, or no usable data)
counts as processed.

```python
from pathlib import Path

LISTING = next(p for p in (Path("processed.txt"), Path("coverage/processed.txt")) if p.exists())


def read_listing(path=LISTING):
    """The processed shot numbers, the listing's (first, last) and its date."""
    head, shots = {}, set()
    for line in path.read_text().splitlines():
        if line.startswith("#"):
            key, _, value = line[1:].strip().partition(" ")
            head[key] = value
        elif line.strip():
            a, _, b = line.strip().partition("-")
            shots.update(range(int(a), int(b or a) + 1))
    first, last = map(int, head["range"].split("-"))
    return shots, first, last, head["updated"]


processed, first, last, updated = read_listing()
total = last - first + 1
print(f"{100 * len(processed) / total:.1f} % of DIII-D shots {first}-{last} processed "
      f"({len(processed):,} of {total:,}; listing of {updated})")
```
